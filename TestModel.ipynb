```python
import pickle
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully!")
```

```python
# Load the saved model pipeline
try:
    with open('musicmood_pipeline.pkl', 'rb') as f:
        pipeline = pickle.load(f)
    
    print("✅ Model pipeline loaded successfully!")
    
except FileNotFoundError:
    print("❌ Error: musicmood_pipeline.pkl not found!")
    print("Please make sure you've run the NB.ipynb notebook first to create the model.")
```

```python
# Load music dataset for recommendation testing
try:
    df_song = pd.read_csv('./dataset/data_moods.csv')
    # Remove 'Calm' mood as done in training
    df_song = df_song[df_song['mood'] != 'Calm']
    
    print("✅ Music dataset loaded successfully!")
    print(f"Dataset shape: {df_song.shape}")
    print(f"Available moods: {df_song['mood'].unique()}")
    print(f"Number of songs per mood:")
    print(df_song['mood'].value_counts())
    
except FileNotFoundError:
    print("❌ Error: Music dataset not found at './dataset/data_moods.csv'")
    print("Music recommendation testing will be limited.")
    df_song = None
except Exception as e:
    print(f"❌ Error loading music dataset: {e}")
    df_song = None
```

```python
def predict_mood(user_text):
    """Predict mood from user input"""
    prediction = pipeline.predict([user_text])[0]
    probabilities = pipeline.predict_proba([user_text])[0]
    classes = pipeline.classes_
    
    confidence = max(probabilities)
    return prediction, confidence

def recommend_songs(mood, top_n=5):
    """Get song recommendations based on mood"""
    if df_song is None:
        return None
    
    recommended = df_song[df_song['mood'] == mood].sort_values(
        by='popularity', ascending=False
    ).head(top_n)[['name', 'artist', 'album']]
    
    return recommended

print("✅ Functions ready!")
```

```python
while True:
    print("\n" + "="*50)
    print("🎵 MUSIC MOOD PREDICTION TEST")
    print("="*50)
    print("1. Test your mood")
    print("2. Exit")
    
    choice = input("\nChoose option (1 or 2): ")
    
    if choice == "1":
        # Get user input
        user_text = input("\n📝 Enter your text: ")
        
        if user_text.strip():
            # Predict mood
            mood, confidence = predict_mood(user_text)
            
            print(f"\n🎯 Predicted Mood: {mood}")
            print(f"📊 Confidence: {confidence:.1%}")
            
            # Get recommendations
            recommendations = recommend_songs(mood)
            
            print(f"\n🎼 Song Recommendations for {mood} mood:")
            if recommendations is not None and not recommendations.empty:
                for i, (_, song) in enumerate(recommendations.iterrows(), 1):
                    print(f"{i}. {song['name']} - {song['artist']}")
            else:
                print("No recommendations available")
        else:
            print("❌ Please enter some text!")
            
    elif choice == "2":
        print("\n👋 Goodbye!")
        break
    else:
        print("❌ Invalid choice! Please choose 1 or 2.")
```
